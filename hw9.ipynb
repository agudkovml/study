Задание №1
Напишите функцию, которая классифицирует фильмы из материалов занятия по правилам:
-оценка 2 и ниже — низкий рейтинг;
-оценка 4 и ниже — средний рейтинг;
-оценка 4.5 и 5 — высокий рейтинг.
Результат классификации запишите в столбец class.

```python
import pandas as pd
```

```python
ratings = pd.read_csv("ratings.csv", encoding="utf-8")
def ratings_classification(rating):
    if rating <= 2.0:
        return "Низкий"
    elif rating <= 4.0:
        return "Средний"
    else:
        return "Высокий"
ratings["class"] = ratings["rating"].apply(ratings_classification)
ratings.head()
```

Задание №2
Напишите гео-классификатор, который каждой строке сможет выставить географическую принадлежность определённому региону. 
Если поисковый запрос не содержит названия города, то ставим ‘undefined’.

```python
geo_data = {
    "Центр": ["москва", "тула", "ярославль"],
    "Северо-Запад": ["петербург", "псков", "мурманск"],
    "Дальний Восток": ["владивосток", "сахалин", "хабаровск"],
}

city = pd.read_csv("city.csv", encoding="utf-8")

def region_classification(city):
    city = str(city).lower().strip()
    if city in geo_data["Центр"]:
        return "Центр"
    elif city in geo_data["Северо-Запад"]:
        return "Северо-Запад"
    elif city in geo_data["Дальний Восток"]:
        return "Дальний Восток"
    else:
        return "undefined"

city["region"] = city["city"].apply(region_classification)
city.head()
```

Задание №3
Проверить, верно ли, что с ростом года выпуска фильма его средний рейтинг становится ниже.
Алгоритм:
1.В переменную years запишите список из всех годов с 1950 по 2010 года.
2.Напишите функцию production_year, которая каждой строке из названия фильма выставляет год выпуска. Не все названия фильмов содержат год выпуска в одинаковом формате, поэтому используйте алгоритм:
    -для каждой строки пройдите по всем годам списка years;
    -если номер года присутствует в названии фильма, то функция возвращает этот год, как год выпуска;
    -если ни один из номеров года списка years не встретился в названии фильма, то возвращается 1900 год.
3.Запишите год выпуска фильма по алгоритму пункта 2 в новый столбец ‘year’.
4.Посчитайте средний рейтинг всех фильмов для каждого значения столбца ‘year’ и отсортируйте результат по убыванию рейтинга.

```python
movies = pd.read_csv("movies.csv", encoding="utf-8")
ratings = pd.read_csv("ratings.csv", encoding="utf-8")
ratings_movies = pd.merge(ratings, movies, on="movieId", how="left")

years = [year for year in range(1950, 2011)]


def production_year(title):
    movie = (title.split()[-1])[1:-1]
    if movie in map(str, years):
        return int(movie)

    return 1900


ratings_movies["year"] = ratings_movies["title"].apply(production_year)
ratings_movies.groupby("year")["rating"].mean().sort_values(ascending=False)
```
