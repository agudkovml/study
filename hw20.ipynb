### Задание 1

Дан объект в $2D$ пространстве

```python
import numpy as np
import matplotlib.pyplot as plt

A = [
[-100, -100, 1],
[-100, 100, 1],
[100, 100, 1],
[0, 0, 1],
[100, -100, 1],
[-100, -100, 1]
]

A = np.array(A)

x = A[:,0]
y = A[:,1]
plt.plot(x, y)
plt.grid(True)
plt.ylim([-200, 200])
plt.xlim([-200, 200]) 
```

При помощи линейных отображений:
- Уменьшить объект в два раза
- Повернуть на 130 градусов
- Отразить объект относительно прямой y=x (поможет матрица `[[0, 1, 0], [1, 0, 0], [0, 0, 1]]`)


```python
decrease = np.array([[0.5, 0, 0], [0, 0.5, 0], [0, 0, 1]])
A2 = A @ decrease
x2 = A2[:,0]
y2 = A2[:,1]
plt.plot(x2, y2)
plt.grid(True)
plt.ylim([-200, 200])
plt.xlim([-200, 200])
```

```python
angle = 130
theta = np.deg2rad(angle)
rotation = np.array([[np.cos(theta), -np.sin(theta), 0], [np.sin(theta), np.cos(theta), 0], [0, 0, 1]])
A3 = A2 @ rotation
x3 = A3[:,0]
y3 = A3[:,1]
plt.plot(x3, y3)
plt.grid(True)
plt.ylim([-200, 200])
plt.xlim([-200, 200])   
```

```python
reflection = np.array([[0, 1, 0], [1, 0, 0], [0, 0, 1]])
A4 = A3 @ reflection
x4 = A4[:,0]
y4 = A4[:,1]
plt.plot(x4, y4)
plt.grid(True)
plt.ylim([-200, 200])
plt.xlim([-200, 200])
```

### Задание 2

Найдите спектральное разложение матрицы:

```python
m = np.array([[1, 2],  [2, 3]])

eigenvalues, Q = np.linalg.eigh(m)
Lambda = np.diag(eigenvalues)
m_back = Q @ Lambda @ Q.T

print("Собственные значения:")
print(eigenvalues)

print("Матрица Q из собственных векторов:")
print(Q)

print("Диагональная матрица Lambda:")
print(Lambda)

print("Восстановленная матрица:")
print(m_back)

print("Проверка:")
print(np.allclose(m, m_back))
```
