## Домашнее задание к занятию "A/B-тесты"

```python
import pandas as pd
from scipy.stats import ttest_ind
from scipy.stats import f_oneway, shapiro
import warnings
warnings.filterwarnings("ignore")

pokemon = pd.read_csv('https://raw.githubusercontent.com/a-milenkin/datasets_for_t-tests/main/pokemon.csv', on_bad_lines='skip') 
pokemon.head(50)

# Обратите внимание, что у покемона может быть один или два класса.
# Если у покемона два класса, считается, что они имеют одинаковую значимость.
```

### Задачи

<div class="alert alert-info">
<b>Задание № 1:</b>
    
Профессор Оук подозревает, что покемоны в классе `Grass` имеют более сильную обычную атаку, чем покемоны в классе `Rock`. Проверьте, прав ли он, и убедите его в своём выводе статистически.
    
    
Примечание: если есть покемоны, которые относятся к обоим классам, просто выбросьте их;
    
Вы можете предположить, что распределение обычных атак является нормальным для всех классов покемонов.

</div>

```python
grass = pokemon[
    ((pokemon['Class 1'] == 'Grass') | (pokemon['Class 2'] == 'Grass')) &
    ~(
        ((pokemon['Class 1'] == 'Rock') & (pokemon['Class 2'] == 'Grass')) |
        ((pokemon['Class 1'] == 'Grass') & (pokemon['Class 2'] == 'Rock'))
    )
]

rock = pokemon[
    ((pokemon['Class 1'] == 'Rock') | (pokemon['Class 2'] == 'Rock')) &
    ~(
        ((pokemon['Class 1'] == 'Rock') & (pokemon['Class 2'] == 'Grass')) |
        ((pokemon['Class 1'] == 'Grass') & (pokemon['Class 2'] == 'Rock'))
    )
]

grass_mean = grass['Attack'].mean()
rock_mean = rock['Attack'].mean()

stat, p = ttest_ind(grass['Attack'], rock['Attack'], equal_var=False)
if p > 0.05:
    print("Различия между группами не значимы")
else:   
    print("Различия между группами значимы")
    if grass_mean > rock_mean:
        print("Группа Grass имеет более высокий средний показатель атаки")
    elif rock_mean > grass_mean:
        print("Группа Rock имеет более высокий средний показатель атаки")
        
#Он был не прав
    
```

<div class="alert alert-info">
<b>Задание № 2:</b>
    
Профессор Оук уже долго не может спать по ночам, ведь его волнует вопрос, а правда ли, что покемоны в классе `Water` в среднем быстрее, чем покемоны в классе `Normal`.
    
    
Проверьте, прав ли он, и убедите его в своём выводе статистически.
    
Примечание: если есть покемоны, которые относятся к обоим классам, выбросьте их;
    
Вы можете предположить, что распределение скорости движения является нормальным для всех классов покемонов.
</div>

```python
water = pokemon[
    ((pokemon['Class 1'] == 'Water') | (pokemon['Class 2'] == 'Water')) &
    ~(
        ((pokemon['Class 1'] == 'Normal') & (pokemon['Class 2'] == 'Water')) |
        ((pokemon['Class 1'] == 'Water') & (pokemon['Class 2'] == 'Normal'))
    )
]

normal = pokemon[
    ((pokemon['Class 1'] == 'Normal') | (pokemon['Class 2'] == 'Normal')) &
    ~(
        ((pokemon['Class 1'] == 'Water') & (pokemon['Class 2'] == 'Normal')) |
        ((pokemon['Class 1'] == 'Normal') & (pokemon['Class 2'] == 'Water'))
    )
]


stat, p = ttest_ind(water['Speed'], normal['Speed'], equal_var=False)
if p > 0.05:
    print("Различия между группами не значимы")
else:   
    print("Различия между группами значимы")
    if water['Speed'].mean() > normal['Speed'].mean():
        print("Группа Water имеет более высокий средний показатель скорости")
    else:
        print("Группа Normal имеет более высокий средний показатель скорости")
        
#Он был не прав
    
```

<div class="alert alert-info">
<b>Задание № 3:</b>
    
Профессор Оук тот еще безумец. Он изобрёл сыворотку, способную ускорить покемона. Однако мы усомнились в эффективности его вакцины. Професоор дал эту сыворотку следующим покемонам: смотри массив `treathed_pokemon`. Проверьте, работает ли вообще его сыворотка, убедите всех в своём выводе статистически.
    
    
Вы можете предположить, что распределение скорости движения является нормальным для всех классов покемонов.

</div>

```python
# Покемоны, которые принимали сыворотку увеличения скорости
treathed_pokemon = ['Mega Beedrill', 'Mega Alakazam',
                    'Deoxys Normal Forme', 'Mega Lopunny']

boost_pokemon = pokemon[pokemon['Name'].isin(treathed_pokemon)]
other_pokemon = pokemon[~pokemon['Name'].isin(treathed_pokemon)]

boost_pokemon['Speed'].mean()
other_pokemon['Speed'].mean()

stat, p = ttest_ind(boost_pokemon['Speed'], other_pokemon['Speed'], equal_var=False)
if p > 0.05:
    print("Различия между группами не значимы")
else:   
    print("Различия между группами значимы")
    if boost_pokemon['Speed'].mean() > other_pokemon['Speed'].mean():
        print("Группа покемонов, принимавших сыворотку, имеет более высокий средний показатель скорости")
    else:
        print("Группа покемонов, не принимавших сыворотку, имеет более высокий средний показатель скорости")
        
#Он был прав
```

<div class="alert alert-info">
<b>Задание № 4:</b>
    
Профессор Оук всегда любил истории про легендарных покемонов. Однако профессор не очень уверен, что они лучше остальных покемонов. Оук предложил разобраться в этом нам. Проверьте, действительно ли сумма характеристик `HP`,`Attack`,`Defense` у легендарных покемонов выше, чем у других покемонов?

А произведение этих же параметров?

Найдите ответы на эти вопросы и убедите всех в своём выводе статистически.
   

Вы можете предположить, что распределение сум и произведений этих параметров является нормальным для всех классов покемонов.

</div>

```python
legendary_pokemons = pokemon[pokemon['Legendary'] == True]
non_legendary_pokemons = pokemon[pokemon['Legendary'] == False]

stat, p = ttest_ind(legendary_pokemons['HP'] + legendary_pokemons['Attack'] + legendary_pokemons['Defense'],
                    non_legendary_pokemons['HP'] + non_legendary_pokemons['Attack'] + non_legendary_pokemons['Defense'], 
                    equal_var=False)
if p > 0.05:
    print("Различия между группами не значимы")
else:   
    print("Различия между группами значимы")
    if (legendary_pokemons['HP'] + legendary_pokemons['Attack'] + legendary_pokemons['Defense']).mean() > (non_legendary_pokemons['HP'] + non_legendary_pokemons['Attack'] + non_legendary_pokemons['Defense']).mean():
        print("Группа легендарных покемонов имеет более высокий средний показатель суммы HP, Attack и Defense")
    else:
        print("Группа не легендарных покемонов имеет более высокий средний показатель суммы HP, Attack и Defense")
        
stat, p = ttest_ind(legendary_pokemons['HP'] * legendary_pokemons['Attack'] * legendary_pokemons['Defense'],
                    non_legendary_pokemons['HP'] * non_legendary_pokemons['Attack'] * non_legendary_pokemons['Defense'], 
                    equal_var=False)

if p > 0.05:
    print("Различия между группами не значимы")
else:   
    print("Различия между группами значимы")
    if (legendary_pokemons['HP'] * legendary_pokemons['Attack'] * legendary_pokemons['Defense']).mean() > (non_legendary_pokemons['HP'] * non_legendary_pokemons['Attack'] * non_legendary_pokemons['Defense']).mean():
        print("Группа легендарных покемонов имеет более высокий средний показатель произведения HP, Attack и Defense")
    else:
        print("Группа не легендарных покемонов имеет более высокий средний показатель произведения HP, Attack и Defense")
        
#Он был прав
```

<div class="alert alert-info">
<b>Задание № 5:</b>
    
Профессор Оук частенько наблюдает за боями покемонов. После очередных таких боёв Оук выделил четыре класса `best_defence_class`, которые на его взгляд одинаковы по "силе обычной защиты" `Defense`.

Проверьте, действительно ли эти классы покемонов не отличаются по уровню защиты статистически значимо? Всё та же статистика вам в помощь!
   

Вы можете предположить, что распределение параметров защитных характеристик является нормальным для всех классов покемонов.

</div>

```python
best_defence_class = ['Rock', 'Ground', 'Steel', 'Ice']
best_defence_pokemons = pokemon[(pokemon['Class 1'].isin(best_defence_class)) | (pokemon['Class 2'].isin(best_defence_class))]
other_pokemons = pokemon[~((pokemon['Class 1'].isin(best_defence_class)) | (pokemon['Class 2'].isin(best_defence_class)))]

stat, p = ttest_ind(best_defence_pokemons['Defense'], other_pokemons['Defense'], equal_var=False)
if p > 0.05:
    print("Различия между группами не значимы")
else:   
    print("Различия между группами значимы")
    if best_defence_pokemons['Defense'].mean() > other_pokemons['Defense'].mean():
        print("Группа покемонов с лучшей защитой имеет более высокий средний показатель Defense")
    else:
        print("Группа покемонов с лучшей защитой имеет более низкий средний показатель Defense")
        
#Он был прав
```
